So moving forward, I'd like to try and do a forecasting model for the alcoholic beverages that was consumed the most across all the regions in Russia. In this case, it's beer.

```python
#Now to implement my time series forcasting model. This model will forecast the mean consumption of beer in this data set for the next ten year.
#First, to import the Augmented Dickey Fuller Test to test for stationarity

from statsmodels.tsa.stattools import adfuller
```

```python
total_consumption_of_beer_per_year = df.groupby(['Year'])["Beer"].sum()
print(total_consumption_of_beer_per_year)
print(total_consumption_of_beer_per_year.shape)
```

```python
total_consumption_of_beer_per_year = total_consumption_of_beer_per_year.diff().dropna()
```

```python
#Now to test for stationarity which is the d value
adf_result_total_consumption_of_beer_per_year = adfuller(total_consumption_of_beer_per_year)
print(f"ADF Statistic: {adf_result_total_consumption_of_beer_per_year[0]}")
print(f"p-value: {adf_result_total_consumption_of_beer_per_year[1]}")
```

Since the p-value (0.000135) is much smaller than the common significance level of 0.05, we can confidently reject the null hypothesis.
The null hypothesis of the Augmented Dickey-Fuller test states that the time series has a unit root and is therefore non-stationary.
Rejecting the null hypothesis means that the total_consumption_of_beer_per_year series is stationary.
